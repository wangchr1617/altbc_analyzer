# ALTBC：周期几何、正确计数与高清 PNG
运行前在当前环境安装 `pip install -e .`。中心 B 为 Ge；以下默认不限邻居元素，纯 Te–Ge–Te 可添加 `neighbor_element="Te"`。
轨迹使用全部 200 帧；图片 600 dpi，统计网格 0.02 Å，无平滑。仅作结构表征，不单凭图像判定电子机制。

```python
from pathlib import Path

from altbc_analyzer import ALTBCAnalyzer

ROOT = Path.cwd()
if not (ROOT / "pyproject.toml").is_file():
    ROOT = ROOT.parent
assert (ROOT / "pyproject.toml").is_file(), "请从仓库根目录或 example 目录打开 Notebook"

analyzer = ALTBCAnalyzer(
    ROOT / "example/test.xyz",
    filetype="MultiSystem",
    frame_interval=1,
    center_element="Ge",
    apply_mic=True,
    grid_size=0.02,
)
fig, ax = analyzer.plot_grid(ROOT / "outputs/trajectory_ge.png", dpi=600)
analyzer.save_grid(ROOT / "outputs/trajectory_ge.npz")
print(analyzer.stats)
```

```python
crystal = ALTBCAnalyzer(
    ROOT / "example/POSCAR", center_element="Ge", apply_mic=True, grid_size=0.02
)
fig, ax = crystal.plot_grid(ROOT / "outputs/poscar_ge.png", dpi=600)
print(crystal.stats)
```

`compute_grid()` 每次清零重算；`plot_grid()` 复用网格且不修改原计数。`normalization="density"` 为窗口内概率密度，`"per_center"` 为按所有帧中心原子总数归一化的面密度；默认 `"max"` 只显示相对强度。详见 README。
